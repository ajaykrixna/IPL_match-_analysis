```python
merged['bowler_runs']=merged['batsman_runs']+merged['wide_runs']+merged['noball_runs']
```

```python
total_runs=merged.groupby('bowler')['bowler_runs'].sum().reset_index()
```

```python
total_balls=merged.groupby('bowler').size().reset_index()
```

```python
total_balls=total_balls[total_balls[0]>=500]
```

```python
merged_bowler=pd.merge(left=total_runs,right=total_balls,left_on='bowler',right_on='bowler',how='inner')
```

```python
merged_bowler
```

```python
merged_bowler['economy']=merged_bowler['bowler_runs']/(merged_bowler[0]/6)
```

```python
merged_bowler
```

```python
merged_bowler=merged_bowler.sort_values('economy',ascending=True).head(10)
```

```python
merged_bowler
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=merged_bowler['bowler'],y=merged_bowler['economy'],data=merged_bowler)
plt.title("Top 10 Economy Bowlers")
plt.xlabel("Bowlers")
plt.ylabel("Economy")
plt.xticks(rotation=45)
plt.show()
```

**How many matches did each team win by batting first vs batting second?**

```python
defending=df[(df['win_by_runs']>0)].groupby('winner').size().sort_values(ascending=False).reset_index()
```

```python
chasing=df[(df['win_by_wickets']>0)].groupby('winner').size().sort_values(ascending=False).reset_index()
```

```python
defending
```

```python
chasing
```

```python
defending['type']= 'Defending'
chasing['type']= 'Chasing'
```

```python
combined=pd.concat([defending,chasing])
```

```python
combined=combined.rename(columns={0:'Wins'})
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=combined['winner'],y=combined['Wins'],data=combined,hue='type')
plt.title("Team Performance")
plt.xlabel("Teams")
plt.ylabel("Wins")
plt.xticks(rotation=45)
plt.show()
```
